# Строим диксретный вариационный ряд(выборка, упорядоченная по возрастанию)

```python
data_cont = [
    4.49, 9.25, 7.94, 9.10, 6.27, 6.77, 3.47, 8.84, 6.48,
    4.92, 6.98, 10.1, 6.32, 6.36, 5.16, 7.92, 12.0, 7.46,
    7.01, 13.0, 7.34, 6.71, 5.48, 9.95, 11.9, 8.89
]

n = len(data_cont)

sorted_data = sorted(data_cont)

print("Вариационный ряд:")
for x in sorted_data:
    print(x, end=" ")
```

# Cтроим интервальный вариационный ряд

```python
from math import log10
import matplotlib.pyplot as plt

xmin = min(data_cont)
xmax = max(data_cont)

m = round(1 + 3.322 * log10(n)) # определяем  число интервалов по формуле стреджесса

h = (xmax - xmin) / m # длина интервала

print("xmin =", xmin)
print("xmax =", xmax)
print("m =", m)
print("h =", h)
```

Запускается цикл, который делит весь диапазон данных на $m$ равных отрезков. Для каждого интервала подсчитывается количество попавших в него элементов (count).
Чтобы не потерять максимальное значение, для всех интервалов, кроме последнего, используется полуоткрытый промежуток $[left; right)$, а для последнего — закрытый $[left; right]$.

```python
intervals = [] #Формирование интервалов и частот

left = xmin #Левая граница первого интервала равна минимальному значению выборки

for i in range(m): # Цикл по всем интервалам- для каждого интервала подсчитывается количество элементов выборки
    right = left + h # Находим правую границу текущего интервала

    count = 0 #Будет показывать, сколько элементов попало в текущий интервал

    for value in data_cont: #Перебираем ВСЕ элементы выборки

        if i != m - 1: #Проверяем: не последний ли это интервал
            if left <= value < right: #Для всех интервалов кроме последнего проверяем: принадлежит ли значение интервалу
                count += 1 #То есть нашли ещё одно значение в данном интервале
        else: #Если интервал последний
            if left <= value <= right: #Здесь правая граница включается
                count += 1

    intervals.append((left, right, count)) #После проверки всех элементов выборки имеем частоту текущего интервала и сохноняем инф об интревале
    left = right #Правая граница текущего интервала становится левой границей следующего

print("Интервальный вариационный ряд:\n")

for interval in intervals:
    l, r, freq = interval
    print(f"[{l:.3f}; {r:.3f}) : {freq}")
```

# Эмпирическая функция распределения
По определению F_n(x) =k/n где k — число элементов выборки, не превосходящих x

```python
x_values = []
y_values = []

for i in range(n): #формируем точки функции
    x_values.append(sorted_data[i])
    y_values.append((i + 1) / n) 

print("Точки эмпирической функции:")

for x, y in zip(x_values, y_values):
    print(f"x={x:.2f}, F(x)={y:.4f}")
```

```python
plt.figure(figsize=(8,5))

plt.step(x_values, y_values, where='post')

plt.title("Эмпирическая функция распределения")
plt.xlabel("x")
plt.ylabel("F*(x)")
plt.grid(True)

plt.show()
```

```python
data_disc = [
    2,0,3,5,1,2,4,5,3,4,4,0,1,5,1,
    1,3,4,0,2,3,1,3,2,1,2,4,2,0,2
]

n = len(data_disc)

print("Объем выборки =", n)
```

# Строим вариационный ряд


```python
sorted_disc = sorted(data_disc)

print("Вариационный ряд:")

for x in sorted_disc:
    print(x, end=" ")
```

Создается список уникальных значений выборки (values).

Во вложенном цикле подсчитывается, сколько раз (count) каждое уникальное число встречается в исходной выборке

Результат: Получается статистическая таблица, где каждому варианту $x_i$ сопоставлена его абсолютная частота $n_i$

```python
values = [] #Таблица частот

for x in sorted_disc: #Перебираем отсортированную выборку
    if x not in values: #Проверяем: есть ли уже такое значение в списке
        values.append(x) #Если нет — добавляем

freqs = [] #Список частот

for value in values: #Берём очередной вариант
    count = 0 #Обнуляем счётчик.
    for x in data_disc: #Просматриваем всю выборку
        if x == value: #Проверяем: равен ли элемент текущему варианту
            count += 1 #если нашли то добавляем 

    freqs.append(count) #Сохраняем частоту

print("x_i    n_i")

for x, n_i in zip(values, freqs):
    print(f"{x:2d}    {n_i:2d}")
```

```python
plt.figure(figsize=(8,5)) 

plt.plot(values, freqs, marker='o')

plt.title("Полигон частот")
plt.xlabel("Значение")
plt.ylabel("Частота")
plt.grid(True)

plt.show()
```

В цикле рассчитывается накопленная частота current. Значение функции в каждой точке определяется как отношение накопленной частоты к общему объему выборки: $F^*(x) = \frac{\sum n_i}{n}$.

```python
emp_x = [] #Эмпирическая функция распределения
emp_y = [] #оздаются списки координат графика

current = 0 #Накопленная частота

for value, freq in zip(values, freqs):

    current += freq #Добавляем текущую частоту

    emp_x.append(value) #охраняем значение случайной величины
    emp_y.append(current / n) #Вычисляем эмпирическую функцию: накопленная частота/n

print("Таблица F*(x):")

for x, y in zip(emp_x, emp_y):
    print(f"x={x}, F(x)={y:.4f}")
```

# Построение интервального ряда

```python
from math import log10 

xmin = min(data) #Для непрерывной величины необходимо разбить диапазон значений на интервалы
xmax = max(data)

m = round(1 + 3.322 * log10(n)) 

h = (xmax - xmin) / m 

print("Количество интервалов =", m)
print("Ширина интервала =", round(h,3))
```

Проходим циклом по всей выборке и для каждого из $m$ интервалов считает, сколько чисел туда попало.

```python
intervals = [] #Вычисление частот
freqs = []

left = xmin

for i in range(m):     #цикл по всем интервалам

    right = left + h   #вычисляем правую границу интервала

    count = 0          # астота текущего интервала

    for x in data:     #просматриваем все элементы выборки

        if i != m - 1: #если интервал не последний

            if left <= x < right:  #проверяем попадание в интервал
                count += 1         # величиваем частоту

        else:                      # оследний интервал

            if left <= x <= right: #включаем правую границу
                count += 1

    intervals.append((left, right)) # сохраняем границы интервала

    freqs.append(count)             # сохраняем частоту

    left = right                    # следующий интервал начинается здесь

print("Интервальный ряд:\n")

for i in range(m):
    l, r = intervals[i]
    print(f"[{l:.2f}; {r:.2f})  {freqs[i]}")
```

```python
centers = []   # середины интервалов

for l, r in intervals:
    centers.append((l + r) / 2)  # вычисляем центр интервала

plt.figure(figsize=(8,5))        # создаем окно графика

plt.bar(
    centers,                    
    freqs,                    
    width=h*0.9                  
)

plt.title("Гистограмма")
plt.xlabel("Интервалы")
plt.ylabel("Частота")

plt.grid(True)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    centers,
    freqs,
    marker='o'
)

plt.title("Полигон частот")
plt.xlabel("Середины интервалов")
plt.ylabel("Частота")

plt.grid(True)

plt.show()
```

Создает список sum_freq, где каждая следующая сумма — это сумма всех предыдущих частот плюс текущая (накопление).

Извлекает правые границы интервалов в список right_bounds.

Рисует график, где по оси $X$ — правые границы, а по оси $Y$ — накопленное количество элементов.

```python
sum_freq = []  # накопленные частоты

s = 0          # текущая сумма частот

for f in freqs:
    s += f               # добавляем очередную частоту
    sum_freq.append(s)   # сохраняем накопленную частоту

right_bounds = []        # правые границы интервалов

for l, r in intervals:
    right_bounds.append(r)

plt.figure(figsize=(8,5))

plt.plot(
    right_bounds,  # правая граница интервала
    sum_freq,      # накопленная частота
    marker='o'
)

plt.title("Огива")
plt.xlabel("Правая граница интервала")
plt.ylabel("Накопленная частота")

plt.grid(True)

plt.show()
```

# Лабораторная работа 3

```python
data = [
    3.48, 7.87, 4.04, 8.81, 4.08, 3.46, 4.37,
    6.66, 1.46, 5.59, 3.78, 8.73, 5.57,
    8.62, 3.25, 3.38, 4.20, 2.49, 6.11,
    4.54, 6.53, 5.20, 3.84, 5.35, 9.72, 4.63
]
```

# Основные характеристики

```python
from math import log10 

n = len(data) # Объем выборки (количество наблюдений)

u_max = max(data)
u_min = min(data) #

R = u_max - u_min # Размах выборки

N = 1 + 3.322 * log10(n) # количества интервалов группировки
N_round = round(N) # Округляем число интервалов до целого значения

delta = R / (N_round - 1) # Вычисляем ширину интервала
delta_round = round(delta, 2) # Округляем ширину интервала

print("Umax =", round(u_max, 2))
print("Umin =", round(u_min, 2))
print("R =", round(R, 2))
print("n =", n)

print("N =", round(N, 3))
print("N округл. =", N_round)

print("Δ =", round(delta, 4))
print("Δ округл. =", delta_round)
```

Начиная от минимального значения выборки (u_min), цикл последовательно добавляет ширину интервала (delta) к текущей левой границе, пока не сформирует нужное количество интервалов

Результат: Список пар (кортежей), где каждая пара — это начало и конец отрезка. Это позволяет структурировать разбросанные значения выборки в упорядоченную последовательность диапазонов

```python
intervals = [] # Список для хранения интервалов

left = u_min # Левая граница первого интервала

for i in range(N_round - 1): # Создаем интервалы

    right = left + delta # Правая граница интервала

    intervals.append((left, right))  # Сохраняем интервал как пару 

    left = right # Следующий интервал начинается там, где закончился предыдущий

for i, interval in enumerate(intervals, start=1): # Вывод построенных интервалов
    print(
        f"{i}: [{interval[0]:.2f}; {interval[1]:.2f})"
    )
```

подсчет частот

```python
frequencies = [] #Подсчет частот

for i, (left, right) in enumerate(intervals): # Перебираем все интервалы

    count = 0 # Частота текущего интервала

    for value in data: # Просматриваем все элементы выборки

        if i != len(intervals)-1: # Для всех интервалов кроме последнего

            if left <= value < right: # Проверяем попадание значения внутрь интервала
                count += 1

        else:  # Для последнего интервала

            if left <= value <= right: # Включаем правую границу, чтобы максимальный элемент обязательно попал в интервал
                count += 1

    frequencies.append(count)  # Сохраняем частоту интервала
```

# Таблица интервального ряда

```python
print("Интервал\t\tЧастота") 

for (left, right), freq in zip(intervals, frequencies):

    print(
        f"[{left:.2f}; {right:.2f})\t{freq}"
    )
```

```python
import matplotlib.pyplot as plt 

centers = [] # Список центров интервалов

for left, right in intervals: # Для каждого интервала вычисляем его середину
    centers.append((left + right) / 2) # Формула центра интервала: (левая граница + правая граница)/2

plt.figure(figsize=(10, 5))

plt.bar(
    centers,
    frequencies,
    width=delta * 0.9
)

plt.title("Гистограмма выборки")
plt.xlabel("Значения")
plt.ylabel("Частота")

plt.grid(True)

plt.show()
```

# Лабораторная работа 4

```python
data = [
    19.9, 23.2, 22.5, 17.4, 21.8, 18.8, 20.2, 19.3,
    18.4, 19.3, 18.7, 19.4, 18.7, 19.4, 18.7, 16.3,
    18.7, 16.3, 18.4, 19.3, 18.8, 20.5, 20.6, 19.4
]
```

# Основные характеристики

```python
n = len(data) 

xmin = min(data) 

xmax = max(data) 

R = xmax - xmin 


print("Объем выборки n =", n)
print("Минимум =", xmin)
print("Максимум =", xmax)
print("Размах =", round(R, 2))
```

# Среднее арифметическое мода меидана 

```python
s = 0 # Переменная для накопления суммы элементов выборки

for x in data: # Перебираем все элементы выборки
    s += x # Добавляем текущий элемент к общей сумме

mean = s / n # Вычисляем среднее арифметическое

print("Среднее =", round(mean, 4))

sorted_data = sorted(data)  # Создаем отсортированную копию выборки

if n % 2 == 0: # Проверяем, является ли объем выборки четным
    median = (
        sorted_data[n//2 - 1] # Левый центральный элемент
        + sorted_data[n//2] # Правый центральный элемент
    ) / 2 # Среднее двух центральных элементов
else: # Если количество элементов нечетное
    median = sorted_data[n//2] # Берем центральный элемент

print("Медиана =", median)
unique_values = []  # Список для хранения уникальных значений выборки

for x in sorted_data: # Перебираем отсортированные данные
    if x not in unique_values: # Если значение еще не встречалось
        unique_values.append(x) # Добавляем его в список уникальных значений

max_freq = 0 # Максимальная найденная частота
mode = None # Значение моды

for value in unique_values: # Перебираем все уникальные значения

    count = 0 # Счетчик частоты текущего значения

    for x in data:# Просматриваем всю выборку
        if x == value: # Если найдено совпадение
            count += 1 # Увеличиваем частоту

    if count > max_freq: # Если частота больше ранее найденной
        max_freq = count # Запоминаем новую максимальную частоту
        mode = value # Запоминаем соответствующее значение как моду

print("Мода =", mode)
print("Частота моды =", max_freq)
```

# Дисперсия

```python
sum_sq = 0 # Переменная для суммы квадратов отклонений
#Выборочная дисперсия
for x in data: # Перебираем все элементы выборки
    sum_sq += (x - mean) ** 2 # Вычисляем квадрат отклонения от среднего и прибавляем его к общей сумме

variance = sum_sq / n # Вычисляем выборочную дисперсию как среднее значение квадратов отклонений

print("Дисперсия =", round(variance, 4))
```

# Среднеквадратическое отклонение

```python
from math import sqrt 

sigma = sqrt(variance) 

print("Среднеквадратическое отклонение =", round(sigma, 4))
```

# Коэффициент вариации

```python
variation = sigma / mean * 100 
# процентное отношение стандартного отклонения к среднему

print("Коэффициент вариации =", round(variation, 2), "%")
```

```python
data = [
    18.8, 20.2, 19.3, 19.9, 23.2, 22.5, 17.4, 21.8, 19.2,
    19.4, 18.7, 16.3, 18.4, 19.3, 18.8, 19.4, 18.7, 16.3,
    20.5, 20.6, 19.4, 18.7, 16.3, 18.4, 19.3, 18.8, 17.8
]
```

```python
from math import sqrt

n = len(data)

# Среднее
mean = sum(data) / n

# Создаем отсортированную по возрастанию копию выборки
sorted_data = sorted(data)

# Медиана
median = sorted_data[n // 2]

# Мода
unique = [] # Список для хранения уникальных значений выборки

for x in sorted_data:  # Перебираем отсортированные данные
    if x not in unique: # Если значение еще не встречалось
        unique.append(x) # Добавляем его в список уникальных значений

mode = None # Переменная для хранения моды
max_freq = 0 # Максимальная найденная частота

for value in unique: # Перебираем все уникальные значения

    freq = 0 # Счетчик частоты текущего значения

    for x in data: # Просматриваем всю выборку
        if x == value: # Если элемент равен текущему значению
            freq += 1 # Увеличиваем частоту

    if freq > max_freq:  # Если найденная частота больше предыдущей максимальной
        max_freq = freq # Запоминаем новую максимальную частоту
        mode = value # Запоминаем значение как моду

# Выборочная дисперсия
s2 = 0 # Переменная для суммы квадратов отклонений

for x in data: # Перебираем все элементы выборки
    s2 += (x - mean) ** 2 # Вычисляем квадрат отклонения от среднего и добавляем его к общей сумме

variance = s2 / (n - 1) # Вычисляем исправленную выборочную дисперсию Делим на n - 1 согласно статистической формуле

# Стандартное отклонение как квадратный корень из дисперсии
std = sqrt(variance) 

# Стандартная ошибка, Показывает точность оценки среднего значения
std_error = std / sqrt(n)

# Асимметрия
skew_sum = 0 

for x in data: # Перебираем все элементы выборки
    skew_sum += ((x - mean) / std) ** 3 # Суммируем кубы стандартизованных отклонений

skewness = (n / ((n - 1) * (n - 2))) * skew_sum # Вычисляем коэффициент асимметрии Характеризует степень симметричности распределения

# Эксцесс
kurt_sum = 0

for x in data: # Перебираем все элементы выборки
    kurt_sum += ((x - mean) / std) ** 4 # Суммируем четвертые степени стандартизованных отклонений

kurtosis = ( # Вычисляем коэффициент эксцесса, # Показывает степень "островершинности" распределения относительно нормального распределения
    (n * (n + 1))
    / ((n - 1) * (n - 2) * (n - 3))
) * kurt_sum - (
    3 * (n - 1) ** 2
    / ((n - 2) * (n - 3))
)

# Остальные характеристики
xmin = min(data) # Находим минимальное значение выборки
xmax = max(data)

range_value = xmax - xmin # Вычисляем размах выборки: разность между максимумом и минимумом

total = sum(data) # Вычисляем сумму всех элементов выборки

# Уровень надежности 95%
confidence95 = 1.96 * std_error # Вычисляем предельную ошибку для доверительного уровня 95%

print("ОПИСАТЕЛЬНАЯ СТАТИСТИКА")
print("-" * 40)

print("Среднее                =", round(mean, 4))
print("Стандартная ошибка     =", round(std_error, 4))
print("Медиана                =", round(median, 4))
print("Мода                   =", round(mode, 4))
print("Стандартное отклонение =", round(std, 4))
print("Дисперсия выборки      =", round(variance, 4))
print("Эксцесс                =", round(kurtosis, 4))
print("Асимметрия             =", round(skewness, 4))
print("Размах                 =", round(range_value, 4))
print("Минимум                =", xmin)
print("Максимум               =", xmax)
print("Сумма                  =", round(total, 4))
print("Счет                   =", n)
print("Уровень надежности 95% =", round(confidence95, 4))
```
