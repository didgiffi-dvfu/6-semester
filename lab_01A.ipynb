# Что такое линейна регрессия 
Линейная регрессия — это способ: предсказывать число (например, цену, рост, время)  на основе других числовых признаков
x — это данные (признаки)
theta — это важность каждого признака

```python
def linear_func(theta: np.ndarray, x: np.ndarray) -> float:
    """Computes linear regression value for sample x.
    Arguments:
        theta: 1d array of float, regression coefficients
        x: 1d array of float, sample to compute value for
    Returns:
        float, linear regression value
    """
    # ~ x * ø просто крч считает скалярное произведение 
    return float(np.dot(theta, x))
    
```

Теперь объектов много.
X — это матрица, где:

каждый столбец — отдельный признак.
Если X размера (m, n), то:
m — число объектов,
n — число признаков.
Нужно для каждой строки посчитать скалярное произведение с theta.

Если X — матрица, а theta — вектор, то все предсказания сразу:
